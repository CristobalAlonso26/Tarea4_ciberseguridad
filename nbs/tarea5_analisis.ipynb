## 1. Identificación de la Organización
**Organización analizada:** FlowiseAI
**Repositorios seleccionados:** Los 5 repositorios públicos con mayor cantidad de estrellas (Flowise, FlowiseDocs, FlowiseChatEmbed, FlowiseEmbedReact, FlowisePy).

```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## 2. Evidencia y Carga de Hallazgos
Se ejecutó la herramienta `gitleaks` sobre el historial completo (todos los commits) de los repositorios seleccionados. A continuación, cargamos los reportes JSON generados para su análisis.

```python
reports_dir = '../data/results/gitleaks_reports'
all_findings = []

if os.path.exists(reports_dir):
    for filename in os.listdir(reports_dir):
        if filename.endswith('_report.json'):
            repo_name = filename.replace('_report.json', '')
            filepath = os.path.join(reports_dir, filename)
            
            with open(filepath, 'r', encoding='utf-8') as f:
                try:
                    data = json.load(f)
                    # Si hay hallazgos (el json no está vacío), los agregamos a la lista global
                    if data:
                        for item in data:
                            item['Repository'] = repo_name
                            all_findings.append(item)
                except json.JSONDecodeError:
                    pass

df = pd.DataFrame(all_findings)
print(f"Total de secretos (hallazgos) encontrados en la organización: {len(df)}")
```

```python
# Mostramos una muestra representativa de los datos encontrados
# NOTA: Omitimos la columna 'Secret' de la previsualización por seguridad y buenas prácticas.
if not df.empty:
    display_cols = ['Repository', 'RuleID', 'File', 'Commit', 'Author', 'Date']
    display_cols = [c for c in display_cols if c in df.columns]
    display(df[display_cols].head())
else:
    print("No se encontraron secretos en los repositorios analizados.")
```

## 3. Análisis Cuantitativo
En esta sección visualizamos cómo se distribuyen los hallazgos.

```python
if not df.empty:
    plt.figure(figsize=(10, 5))
    repo_counts = df['Repository'].value_counts()
    sns.barplot(x=repo_counts.index, y=repo_counts.values, palette="viridis")
    plt.title('Cantidad de Secretos Detectados por Repositorio', fontsize=14)
    plt.xlabel('Repositorio')
    plt.ylabel('Cantidad de Hallazgos')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print("No hay datos para graficar.")
```

```python
if not df.empty:
    plt.figure(figsize=(10, 5))
    rule_counts = df['RuleID'].value_counts()
    sns.barplot(x=rule_counts.values, y=rule_counts.index, palette="magma")
    plt.title('Tipos de Secretos (Reglas) Más Frecuentes', fontsize=14)
    plt.xlabel('Cantidad de Hallazgos')
    plt.ylabel('Tipo de Secreto (RuleID)')
    plt.tight_layout()
    plt.show()
```
